```python
import pandas as pd
import numpy as np
```

```python


# Original dataset
df1 = pd.read_csv(
    r"D:\internship-credibility-analyzer\data\fake_job_postings.csv\fake_job_postings.csv"
)

# Convert labels
# Original dataset:
# fraudulent = 1 => fake
# fraudulent = 0 => genuine
df1["label"] = 1 - df1["fraudulent"]

# Create unified text column
df1["posting_text"] = (
    df1["title"].fillna("") + " " +
    df1["company_profile"].fillna("") + " " +
    df1["description"].fillna("") + " " +
    df1["requirements"].fillna("") + " " +
    df1["benefits"].fillna("")
)

df1 = df1[["posting_text", "label"]]

# Indian internship dataset
df2 = pd.read_csv(
    r"D:\internship-credibility-analyzer\data\indian_internship_dataset_1000.csv"
)

df2 = df2[["posting_text", "label"]]

# Merge both datasets
df = pd.concat([df1, df2], ignore_index=True)

print(df.shape)
df.head()
```
